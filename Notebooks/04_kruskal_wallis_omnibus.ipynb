# 04 - Kruskal-Wallis Omnibus Test
Run Kruskal-Wallis H-tests across language groups and calculate epsilon-squared effect sizes.

```python
from scipy import stats
import os
import os
import pandas as pd

# Run from this notebook's folder. The bundled CSV is ../data/.
data_path = os.path.join("..", "data", "final_analysis_dataset.csv")
if not os.path.exists(data_path):
    data_path = "/mnt/data/final_analysis_dataset.csv"
df = pd.read_csv(data_path)
print(f"Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns")
metrics = ['TCR', 'TCI', 'TMR']
languages = df['Language'].dropna().unique()
N, k = len(df), len(languages)
rows = []
for m in metrics:
    groups = [df.loc[df['Language'] == lang, m].dropna().values for lang in languages]
    H, p = stats.kruskal(*groups)
    eps_sq = (H - k + 1) / (N - k)
    rows.append({'Metric':m,'H_statistic':H,'df':k-1,'p_value':p,'epsilon_squared':eps_sq})
kw_df = pd.DataFrame(rows).round(5)
os.makedirs("../outputs/tables", exist_ok=True)
kw_df.to_csv("../outputs/tables/kruskal_wallis_results.csv", index=False)
kw_df
```
