# 10 - Joint Diagnostic Profile
Generate a synthesis profile integrating standardized language means of TCR, TMR, and TCI.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd

# Run from this notebook's folder. The bundled CSV is ../data/.
data_path = os.path.join("..", "data", "final_analysis_dataset.csv")
if not os.path.exists(data_path):
    data_path = "/mnt/data/final_analysis_dataset.csv"
df = pd.read_csv(data_path)
print(f"Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns")

order = ['English', 'Russian', 'Igbo', 'Amharic']
palette = {'English':'#2b5c8f','Russian':'#3d85c6','Igbo':'#e69138','Amharic':'#cc0000'}
metrics = ['TCR', 'TMR', 'TCI']
means = df.groupby('Language')[metrics].mean().reindex(order)
z_means = (means - means.mean()) / means.std()
fig, ax = plt.subplots(figsize=(8, 5.5), dpi=300)
for lang in order:
    ax.plot(metrics, z_means.loc[lang], marker='o', linewidth=2.5, markersize=8, label=lang, color=palette[lang])
ax.axhline(0, color='gray', linestyle=':', alpha=.7)
ax.set_title('Figure 4: Joint Typological Diagnostic Profile (Standardized Z-Scores)', fontsize=12, pad=12, fontweight='bold')
ax.set_xlabel('Diagnostic Dimension', fontsize=11)
ax.set_ylabel('Standardized Relative Disparity (Z-Score)', fontsize=11)
ax.legend(frameon=True, facecolor='white', framealpha=.9)
ax.grid(True, linestyle='--', alpha=.5)
plt.tight_layout()
os.makedirs('../outputs/figures', exist_ok=True)
plt.savefig('../outputs/figures/Figure4_Joint_Profile.jpg', format='jpeg', dpi=300)
plt.show()
print('Saved ../outputs/figures/Figure4_Joint_Profile.jpg')
```
