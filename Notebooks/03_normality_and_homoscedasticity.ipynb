# 03 - Normality and Homoscedasticity Tests
Assess normality with Shapiro-Wilk and equal variances with median-centered Levene tests.

```python
from scipy import stats
import os
import os
import pandas as pd

# Run from this notebook's folder. The bundled CSV is ../data/.
data_path = os.path.join("..", "data", "final_analysis_dataset.csv")
if not os.path.exists(data_path):
    data_path = "/mnt/data/final_analysis_dataset.csv"
df = pd.read_csv(data_path)
print(f"Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns")
metrics = ['TCR', 'TCI', 'TMR']
languages = df['Language'].dropna().unique()
shapiro_results = []
for m in metrics:
    for lang in languages:
        values = df.loc[df['Language'] == lang, m].dropna()
        stat, p = (1.0, 1.0) if values.nunique() <= 1 else stats.shapiro(values)
        shapiro_results.append({'Metric':m,'Language':lang,'W_stat':stat,'p_value':p,'Normal_p_gt_05':p > .05})
shapiro_df = pd.DataFrame(shapiro_results)
shapiro_df
```

```python
levene_results = []
for m in metrics:
    groups = [df.loc[df['Language'] == lang, m].dropna().values for lang in languages]
    stat, p = stats.levene(*groups, center='median')
    levene_results.append({'Metric':m,'Levene_W':stat,'p_value':p,'Equal_Variance_p_gt_05':p > .05})
levene_df = pd.DataFrame(levene_results)
os.makedirs("../outputs/tables", exist_ok=True)
shapiro_df.to_csv("../outputs/tables/normality_shapiro_wilk.csv", index=False)
levene_df.to_csv("../outputs/tables/homoscedasticity_levene.csv", index=False)
levene_df
```
