# 01 - Data Validation and Cleaning
**Project:** Cross-Linguistic Tokenization Fairness Study

Load the raw dataset, validate its schema, check missing values and sample balance, and export a cleaned copy.

```python
import os
import pandas as pd

# Run from this notebook's folder. The bundled CSV is ../data/.
data_path = os.path.join("..", "data", "final_analysis_dataset.csv")
if not os.path.exists(data_path):
    data_path = "/mnt/data/final_analysis_dataset.csv"
df = pd.read_csv(data_path)
print(f"Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns")

expected_cols = ['Language', 'Item_ID', 'TCR', 'TCI', 'TMR']
missing_cols = [c for c in expected_cols if c not in df.columns]
assert not missing_cols, f"Missing expected columns: {missing_cols}"
print("--- Missing values ---")
print(df[expected_cols].isnull().sum())
print("\n--- Sample counts per language ---")
counts = df['Language'].value_counts().sort_index()
print(counts)
assert (counts == 75).all(), "Imbalanced language groups detected!"
assert df[expected_cols].notna().all().all(), "Missing values detected in expected columns"
```

```python
os.makedirs("../outputs/tables", exist_ok=True)
df.to_csv("../outputs/tables/cleaned_dataset.csv", index=False)
print("Data validation passed; cleaned dataset saved.")
```
